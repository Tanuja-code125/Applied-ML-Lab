```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report
```

```python

df=pd.read_csv("/Iris-selected-columns.csv")
print(df)
print("\nShape:", df.shape)
print("\nHead:\n", df.head())
print("\n species:\n",df.columns)
print("\nDescription:\n", df.describe())
print("\nInfo:")
df.info()
```

```python
df.columns = ['sepal length (cm)','sepal width (cm)','petal length (cm)','petal width (cm)','species']
print(df.head())
```

```python
plt.scatter(df["sepal length (cm)"], df["sepal width (cm)"],color='red')
plt.scatter(df["petal length (cm)"], df["petal width (cm)"],color='blue')
plt.title('Iris')
plt.xlabel("Length")
plt.ylabel("Width")
plt.tight_layout()
plt.show()
```

```python
X = df[['sepal length (cm)','sepal width (cm)','petal length (cm)','petal width (cm)']]   # Feature (kept as 2D DataFrame for sklearn)
y = df['species']     # Target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
print("X Training data\n",X_train.head())
print("Y training Data\n",y_train.head())
```

```python
model=LogisticRegression()
trained_model=model.fit(X_train,y_train)
print("\n Model trained Succesfullt!")
```

```python
pred=trained_model.predict(X_test)
print("\n Prediction:\n",pred)
```

```python
# Evaluate the model's performance
print("\nClassification Report:\n", classification_report(y_test, pred))
```

```python
print("\nConfusion Matrix:\n", confusion_matrix(y_test, pred))
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, pred)
print(f"Accuracy on testing data: {accuracy:.2f}")
```
